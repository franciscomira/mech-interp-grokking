```
from google.colab import drive
drive.mount('/content/drive')
import os
PROJECT_DIR = '/content/drive/MyDrive/Interpretability'
os.makedirs(PROJECT_DIR + '/checkpoints', exist_ok=True)
print('Project dir:', PROJECT_DIR)
```

```
from transformer_lens import HookedTransformer
device = 'cuda' if torch.cuda.is_available() else 'cpu'
model = HookedTransformer.from_pretrained('attn-only-1l', device=device)
print(model.cfg)
tokens = model.to_tokens('The quick brown fox jumps over the lazy dog')
logits, cache = model.run_with_cache(tokens)
print('logits shape:', logits.shape)
print('cached activations:', len(cache))
```

```
import plotly.express as px
px.imshow(cache['pattern', 0][0, 0].cpu(), title='Layer 0, head 0 attention pattern').show()
print('All good.')
```
